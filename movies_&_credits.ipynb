```python
def convert(text) :
    L = []
    for i in ast.literal_eval(text) :
        L.append(i['name'])
    return L    
```

```python
def convert_t(text) :
    L = []
    count = 0
    for i in ast.literal_eval(text) :
        if(count < 3) :
          L.append(i['name'])
        count += 1
    return L    
```

```python
def fetch_director(text) :
    L = []
    for i in ast.literal_eval(text) :
        if i['job'] == 'Director' :
            L.append(i['name'])
    return L    
```

#### So why the word-to-vec embedding technique didn't work  :

     1. Semantic Bleaching Effect (Zipf's Law)  -- Due to the simple averaging the embedding vectors all the keywords are generally given equal weightage thus the importance of some specific keywords like a cast or crew or some special note point about a movie is not captured

     2. The Hubness Problem (Curse of Dimensionality) -- Since we are only averaging the embedding vectors of such huge dimensions for a small vocabulary the net vectors which will we get for each movie/doc will be such that they would be shifted towards a common hyper centre as close as possible so automatically their cosine similarity will get enhanced even their own semantic meanings are vastly different.
     
            Thats the reason we got the movie matrix list with min cosine similarity of 0.85 even they are very different in terms of meanings

     3.     Loss of Compositionality  -- This is one of the core problems in word2vec or embeddings technique. Word2Vec is excellent at capturing local context (what words appear next to each other in a sentence). However, when you extract those word vectors and average them, you instantly destroy all word order and syntactic relationships. You reduce a highly contextual neural embedding back into a naive "Bag of Words."

      For example, averaging the vectors for the sentence "the alien attacks the human" yields the exact same mathematical result as "the human attacks the alien".        

#### Why doc-to-vec technique didn't work ?

  However, this technique can semantically and syntactically capture the meanings of a doc unlike word2vec as it maintains a separate embedding vector for the doc along with the word but it requires a huge amount of textual data in each doc to effectively capture those meanings. In this data the doc has very less no of words than the actual requirement to train the doc2vec model so it fails to do so here.
